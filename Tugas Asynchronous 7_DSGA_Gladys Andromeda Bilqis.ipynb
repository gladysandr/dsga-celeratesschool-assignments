### **Article Text Summarization**

```python
article = """
## Dampak Ganda E-commerce di Indonesia: Antara Pertumbuhan Ekonomi dan Kesenjangan Digital

### Pendahuluan
Ekonomi digital Indonesia, didominasi oleh sektor e-commerce, telah menjelma menjadi lokomotif utama
pertumbuhan nasional. Studi e-Conomy SEA memproyeksikan nilai ekonomi digital Indonesia akan mencapai
USD 77 miliar pada tahun 2022 dan terus melesat hingga mendekati USD 130 miliar pada tahun 2025.
Perdagangan elektronik atau e-commerce menyumbang porsi terbesar dari angka fantastis ini. Namun,
di balik angka pertumbuhan yang memukau, muncul tantangan serius terkait pemerataan dan kesiapan
sumber daya manusia (SDM) yang menciptakan dampak ganda pada perekonomian.

### Pendorong Pertumbuhan dan Akses Pasar
Kehadiran e-commerce memberikan dampak positif yang fundamental, terutama bagi sektor UMKM. Platform
seperti marketplace telah menghapus batasan geografis, memungkinkan produk dari pengusaha di daerah
terpencil sekalipun untuk diakses oleh konsumen nasional hingga global.

- Meningkatkan Konsumsi Domestik: Kemudahan transaksi, promosi yang intensif, dan variasi produk yang
  luas melalui e-commerce secara langsung meningkatkan daya beli dan konsumsi domestik, yang merupakan
  komponen penting dalam Produk Domestik Bruto (PDB).
- Penciptaan Lapangan Kerja Baru: Pertumbuhan ekosistem e-commerce secara masif menciptakan kebutuhan
  akan profesi baru, mulai dari kurir logistik, content creator, analis data, hingga prompt engineer.
- Efisiensi Biaya Operasional: Bagi pelaku usaha, biaya pemasaran dan distribusi menjadi lebih efisien.
  Mereka tidak lagi memerlukan toko fisik di lokasi mahal.

### Tantangan dan Ancaman Kesenjangan
Meskipun membawa berkah, fenomena e-commerce juga mempertajam masalah struktural dalam perekonomian
Indonesia, khususnya terkait kesiapan digital.

- Kesenjangan Akses Infrastruktur: Pertumbuhan e-commerce sangat timpang antara wilayah perkotaan
  (Jawa-Bali) dan luar Jawa. Infrastruktur internet yang belum merata dan biaya logistik yang tinggi
  di daerah terpencil menjadi penghalang utama.
- Krisis Talenta Digital: Menurut data Kementerian Kominfo, Indonesia membutuhkan sekitar 600.000
  talenta digital baru setiap tahunnya.
- Isu Keamanan dan Regulasi: Peningkatan transaksi juga diikuti dengan peningkatan risiko penipuan
  daring (cybercrime) dan isu perlindungan data pribadi konsumen.
- Pergeseran Perilaku Konsumtif: Kemudahan belanja 24/7 dan flash sale memicu perilaku konsumtif
  yang berlebihan (impulsive buying).

### Kesimpulan
E-commerce di Indonesia berada di persimpangan jalan. Keberlanjutan dan dampaknya yang inklusif sangat
bergantung pada kemampuan negara dalam mengatasi kesenjangan digital, mempercepat pemerataan
infrastruktur, dan mengisi defisit kebutuhan talenta digital melalui reformasi pendidikan dan pelatihan.
Tanpa intervensi yang tepat, e-commerce hanya akan memperkaya wilayah yang sudah maju dan meninggalkan
mereka yang kurang siap.
"""
```

```python
prompt = """
Summarize this article into 5-7 comprehensive and synthesized bullet points.
Do not use sub-headings. Each point should merge related ideas (e.g., combining growth drivers into one point, and infrastructure/talent challenges into another).

Rules:
1. Use highly academic and sophisticated English.
2. Ensure it is a high-level summary, not just a translation of sections.
3. Total word count must be under 500 words.
4. No introductory or closing text."
"""

response = client.models.generate_content(
    model="gemini-2.5-flash",
    contents=f"{prompt}\nArticle: {article}"
)

display(Markdown(response.text))
```
